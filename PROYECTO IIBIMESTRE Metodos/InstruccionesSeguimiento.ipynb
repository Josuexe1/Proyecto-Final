# Objetivo del Proyecto
Simular una impresora 3D que recorre la superficie de un objeto bidimensional representado en archivos SVG, capa por capa, para cubrir el modelo.

## Pasos a Realizar:
1. **Cargar los Archivos SVG**
   - Abre y lee los archivos .svg ubicados en la carpeta `models`. Estos contienen los modelos bidimensionales que la impresora debe simular imprimir.
   - Usa una biblioteca como `matplotlib` (o similar) en Python para visualizar y procesar los modelos.

2. **Crear una Interfaz de Usuario**
   - Desarrolla una interfaz gráfica donde el usuario pueda:
     - Configurar la resolución de impresión (distancia entre puntos de deposición).
     - Establecer la velocidad de la impresión.
   - Herramientas sugeridas: `tkinter`, `PyQt`, o `Dash`.

3. **Calcular la Trayectoria**
   - Diseña un algoritmo que determine la trayectoria de la impresora:
     - Divide el área dentro del modelo SVG en una malla de puntos según la resolución.
     - Recorre la malla capa por capa en orden lógico (de abajo hacia arriba).
     - Optimiza la trayectoria para evitar movimientos innecesarios.

4. **Dibujar la Trayectoria**
   - Implementa una función que grafique:
     - El contorno del modelo SVG.
     - La trayectoria de la impresora según los parámetros configurados.

5. **Preguntas de Análisis**
   - Responde estas preguntas como parte del proyecto:
     - ¿Qué área del objeto es cubierta por la impresora 3D según tu algoritmo?
     - ¿Cómo afecta la resolución configurada al resultado y al rendimiento?
     - ¿En qué casos el algoritmo diseñado no tiene un buen rendimiento?

Para generar la trayectoria capa por capa y adaptarla al modelo SVG, debemos considerar lo siguiente:

- Definir una resolución para dividir el área del contorno SVG.
- Llenar el área dentro del contorno con una malla de puntos.
- Recorrer los puntos de la malla siguiendo un patrón de impresión lógico (como de izquierda a derecha, de abajo hacia arriba).

## Explicación del Código

### Ordenación:
- Los puntos se agrupan por coordenadas Y para determinar las "líneas" horizontales de impresión.
- Dentro de cada línea, los puntos se ordenan por X (posición horizontal).

### Patrón zigzag:
- Se alterna el orden de las líneas:
  - Una línea se recorre de izquierda a derecha.
  - La siguiente línea se recorre de derecha a izquierda.

### Graficar la trayectoria:
- La trayectoria se grafica en verde como una línea continua.
- Los puntos de la trayectoria se grafican como puntos naranjas.

### Resultado
Al ejecutar este código, obtendrás:
- El contorno del modelo en azul.
- La trayectoria de impresión (en zigzag) en verde.
- Los puntos específicos de la trayectoria en naranja.
